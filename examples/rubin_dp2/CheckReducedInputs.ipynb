```python
anacal_input = tables_io.read('8006/anacal_merged_8006.parq')
anacal_reduced = tables_io.read('8006/anacal_reduced_8006.parq')
anacal_input.index = anacal_input.object_id
anacal_reduced.index = anacal_reduced.id
```
