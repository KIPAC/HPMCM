# Check reduced inputs

Validates that reducing input tables preserves coordinates by comparing RA/Dec before and after reduction for Anacal, Object, and Shear catalogs.

```python
# Standard imports
import matplotlib.pyplot as plt
import numpy as np
import tables_io
```

## Configuration

```python
tract = 8006
```

```python
# Load raw and reduced Anacal tables
anacal_input = tables_io.read(f'{tract}/anacal_merged_{tract}.parq')
anacal_reduced = tables_io.read(f'{tract}/anacal_reduced_{tract}.parq')
anacal_input.index = anacal_input.object_id
anacal_reduced.index = anacal_reduced.id
```

```python
# Match on id to compare raw vs reduced
anacal_joined = anacal_input.join(anacal_reduced, how='left', lsuffix='_in', rsuffix='_red')
```

```python
# Check RA/Dec round-trip error for Anacal
ra_diff = anacal_joined.ra_in - anacal_joined.ra_red
dec_diff = anacal_joined.dec_in - anacal_joined.dec_red
print(f"RA {ra_diff.min()} - {ra_diff.max()}")
print(f"DEC {dec_diff.min()} - {dec_diff.max()}")
```

```python
# Load raw and reduced Object catalog tables
object_in = tables_io.read(f'{tract}/object_{tract}.parq')
object_red = tables_io.read(f'{tract}/object_reduced_{tract}.parq')
object_in.index = object_in.objectId
object_red.index = object_red.id
```

```python
object_joined = object_in.join(object_red, how='left', lsuffix='_in', rsuffix='_red')
```

```python
ra_diff = object_joined.coord_ra_in - object_joined.ra
dec_diff = object_joined.coord_dec_in - object_joined.dec
print(f"RA {ra_diff.min()} - {ra_diff.max()}")
print(f"DEC {dec_diff.min()} - {dec_diff.max()}")
```

```python
# Load raw and reduced Rubin shear tables
shear_rubin_in = tables_io.read(f'{tract}/shear_object_{tract}.parq')
shear_rubin_ns = tables_io.read(f'{tract}/shear_object_reduced_uncleaned_{tract}_ns.parq')
shear_rubin_1m = tables_io.read(f'{tract}/shear_object_reduced_uncleaned_{tract}_1m.parq')
shear_rubin_1p = tables_io.read(f'{tract}/shear_object_reduced_uncleaned_{tract}_1p.parq')
shear_rubin_2m = tables_io.read(f'{tract}/shear_object_reduced_uncleaned_{tract}_2m.parq')
shear_rubin_2p = tables_io.read(f'{tract}/shear_object_reduced_uncleaned_{tract}_2p.parq')    
```

```python
shear_rubin_1p.columns
```

```python
shear_rubin_in.index = shear_rubin_in.shearObjectId
shear_rubin_ns.index = shear_rubin_ns.id
shear_rubin_1m.index = shear_rubin_1m.id
shear_rubin_1p.index = shear_rubin_1p.id
shear_rubin_2m.index = shear_rubin_2m.id
shear_rubin_2p.index = shear_rubin_2p.id
```

```python
shear_rubin_in_ns = shear_rubin_in.join(shear_rubin_ns, how='right', lsuffix='_in', rsuffix='_ns')
```

```python
ra_diff = shear_rubin_in_ns.ra_in - shear_rubin_in_ns.ra_ns
dec_diff = shear_rubin_in_ns.dec_in - shear_rubin_in_ns.dec_ns
print(f"RA {ra_diff.min()} - {ra_diff.max()}")
print(f"DEC {dec_diff.min()} - {dec_diff.max()}")
```
