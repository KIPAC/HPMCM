```python
import matplotlib.pyplot as plt
import tables_io
import numpy as np
```

```python
t = tables_io.read('8006/obj_match_object_stats.pq')
```

```python
(np.bitwise_and(t['catalog_mask'], 3) == 3).sum()
```

```python
clusters = tables_io.read('8006/obj_match_cluster_stats.pq')
bins = np.logspace(0, 5, 51)
for i_bit, label in enumerate(['obj', 'anacal', 'rubin_md', 'desc_md']):
    mask_val = 1 << i_bit
    mask = np.bitwise_and(clusters.catalog_mask, mask_val) == mask_val
    _ = plt.hist(clusters.snr[mask], bins=bins, alpha=0.2, label=label)
_ = plt.xscale('log')
_ = plt.yscale('log')
_ = plt.legend()
```

```python
clusters = tables_io.read('8006/obj_match_cluster_stats.pq')
bins = np.logspace(0, 5, 51)
for i_bit, label in enumerate(['obj', 'anacal', 'rubin_md', 'desc_md']):
    mask_val = 1 << i_bit
    mask = np.bitwise_and(clusters.catalog_mask, mask_val) == mask_val
    _ = plt.scatter(clusters.x_pix[mask][::10], clusters.y_pix[mask][::10], s=1, alpha=0.2, label=label)
_ = plt.legend()
```

```python
_ = plt.hist2d(clusters.catalog_mask, clusters.snr, norm='log', bins=(np.linspace(0.5, 15.5, 16), np.logspace(0, 5, 51)))
_ = plt.yscale('log')
```
