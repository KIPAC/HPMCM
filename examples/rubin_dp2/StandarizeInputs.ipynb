# Standarize the input formats for various catalog

```python
# Standard imports
import tables_io
import hpmcm
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Tract number and cell geometry for this dataset
# Set the tract
tract = 8006
# This will need to change with tract
our_geom = hpmcm.shear_utils.ShearCellGeometry(ref_dir=(340.5000000000, -9.6694214876))
```

```python
# Flags controlling which steps to re-run
redo_anacal = False
redo_object = False
redo_rubin_md = False
redo_desc_md = False

# Input file lists
# Files and catalogs ids for rubin metadetect
rubin_shear_match_files = [
    f'{tract}/shear_object_reduced_uncleaned_{tract}_ns.parq',
    f'{tract}/shear_object_reduced_uncleaned_{tract}_1m.parq',
    f'{tract}/shear_object_reduced_uncleaned_{tract}_1p.parq',
    f'{tract}/shear_object_reduced_uncleaned_{tract}_2m.parq',
    f'{tract}/shear_object_reduced_uncleaned_{tract}_2p.parq',    
]
rubin_shear_match_ids = list(range(len(rubin_shear_match_files)))

# Files and catalogs ids for desc metadetect
desc_shear_match_files = [
    f'{tract}/mdet-run-reduced_uncleaned_{tract}_ns.parq',
    f'{tract}/mdet-run-reduced_uncleaned_{tract}_1m.parq',
    f'{tract}/mdet-run-reduced_uncleaned_{tract}_1p.parq',
]
desc_shear_match_ids = list(range(len(desc_shear_match_files)))

# Files and catalogs ids for meta-matching
all_files = [
    f'{tract}/object_reduced_{tract}.parq',
    f'{tract}/anacal_reduced_{tract}.parq',
    f'{tract}/desc_md_match_object_stats.pq',
    f'{tract}/rubin_md_match_object_stats.pq',
]
all_ids = list(range(len(all_files)))

if redo_anacal:
    hpmcm.utils.reduceAnacalTable(
        f'{tract}/anacal_merged_{tract}.parq', 
        f'{tract}/anacal_reduced_{tract}.parq',
    )

if redo_object:
    hpmcm.utils.reduceObjectTable(
        f'{tract}/object_{tract}.parq', 
        f'{tract}/object_reduced_{tract}.parq',
    )

if redo_rubin_md:
    hpmcm.shear_utils.splitRubinMDByTypeAndClean(
        f'{tract}/shear_object_{tract}.parq',
        tract,
        0.01,
        deshear=False,
    )

    for st in ['ns', '1m', '1p', '2m', '2p']:
        hpmcm.utils.reduceRubinMDTable(
            f"{tract}/shear_object_{tract}_uncleaned_{tract}_{st}.parq",
            f"{tract}/shear_object_reduced_uncleaned_{tract}_{st}.parq",
        )

if redo_desc_md:
    hpmcm.shear_utils.splitDESCMDByTypeAndClean(
        f'{tract}/mdet-run-{tract}.parq',
        tract,
        deshear=False,
    )

    for st in ['ns', '1m', '1p']:
        hpmcm.utils.reduceDESCMDTable(
            f"{tract}/mdet-run-{tract}_uncleaned_{tract}_{st}.parq",
            f"{tract}/mdet-run-reduced_uncleaned_{tract}_{st}.parq",
        )
```
