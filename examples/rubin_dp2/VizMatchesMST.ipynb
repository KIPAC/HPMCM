# Visualize MST matches

Loads saved MST match objects and displays cluster structure using the
MST-based visualization utilities.

`showMSTCluster` draws the minimum spanning tree of sources in a cluster,
coloured by catalog ID.

`showMSTObjects` draws the same tree with intra-object edges coloured
solid by object and inter-object (cut) edges dashed grey, making the
splitting decisions visible.

```python
# Standard imports
import tables_io
import hpmcm
import matplotlib.pyplot as plt
import numpy as np
```

## Configuration

```python
tract = 8006
```

## Rubin shear match

```python
# Load saved Rubin shear MST match
rubin_shear_match = hpmcm.MSTShearMatch.load(
    f"{tract}/rubin_mst_md_match",
    (100, 101),
    (100, 101),
)
```

```python
# Cluster view: sources coloured by catalog ID, MST edges in grey
cell_data = rubin_shear_match.cell_dict[20200]
cluster_id = list(cell_data.cluster_dict.keys())[0]
cluster_data = cell_data.cluster_dict[cluster_id]
_ = hpmcm.viz_utils.showMSTCluster(cluster_data, cell_data)
```

```python
# Object view: intra-object edges coloured by object, cut edges dashed grey
_ = hpmcm.viz_utils.showMSTObjects(cluster_data, cell_data)
```
