# Merge per-patch metadect files into a per-tract file

This does not do de-deduplication

```python
# standard imports
import tables_io
import numpy as np
```

```python
# Set the tract 
tract = 8006
```

```python
# loop on patches, read the per-patch file and append it to a list
# Add the patch coordinates as we go
l_cat = []
for i in range(100):
    fname = f"{tract:05}/{tract:05}-{i:05}-mdet.parquet"
    tx = tables_io.read(fname, fmt='pq')
    patch_x = i % 10
    patch_y = i // 10
    tx['patch_x'] = np.full(len(tx), patch_x)
    tx['patch_y'] = np.full(len(tx), patch_y)    
    l_cat.append(tx)
```

```python
# Use tables IO to concatenate the files
# Add an id, for now this is just the ordinal number
md_desc_np = tables_io.concat_table(l_cat, tables_io.types.PD_DATAFRAME)
md_desc_np['id'] = np.arange(len(md_desc_np))
```

```python
# Write the per-tract file
tables_io.write(md_desc_np, f'{tract}/mdet-run-{tract}.parq')
```
