# Download and untar test data

Downloads `8006_inputs.tgz` from the SLAC S3DF store and unpacks it into this directory (`examples/rubin_dp2/`).

```python
import tarfile
import urllib.request
from pathlib import Path

URL = "https://s3df.slac.stanford.edu/people/echarles/package_test_data/hpmcm/8006_inputs.tgz"
DEST_DIR = Path(".")
ARCHIVE = Path("8006_inputs.tgz")
```

```python
print(f"Downloading {URL} ...")
urllib.request.urlretrieve(URL, ARCHIVE)
print(f"Saved to {ARCHIVE} ({ARCHIVE.stat().st_size / 1e6:.1f} MB)")
```

```python
print(f"Extracting {ARCHIVE} -> {DEST_DIR} ...")
with tarfile.open(ARCHIVE) as tf:
    tf.extractall(DEST_DIR)
print("Done.")
```

```python
print("Extracted files:")
for p in sorted(DEST_DIR.rglob("*")):
    if p.is_file():
        print(" ", p)
```
